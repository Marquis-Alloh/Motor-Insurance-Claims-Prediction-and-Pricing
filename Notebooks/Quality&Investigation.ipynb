```python
df_plotting['predicted_missingness'] = nb_model.predict(df_plotting)

# 3. Sort by 'x' (Policy Age), NOT 'y' (Missingness) to fix the zig-zag!
df_sorted = df_plotting.sort_values('x')

# 4. Plot
plt.figure(figsize=(10, 6))

sns.set_theme(style="darkgrid")

# Scatter plot: Policy Age ('x') on the X-axis, Missingness ('y') on the Y-axis
sns.scatterplot(data=df_plotting, x='x', y='y', color='royalblue', s=80, label='Data Points')

# Plot the line of best fit using the SORTED data
plt.plot(df_sorted['x'], df_sorted['predicted_missingness'], 
         color='red', linewidth=2.5, label='Negative Binomial Fit')

# Formatting
plt.title(f'Policy age vs Missingness, given {title}, with a negative binomial fit', fontsize=14, pad=15)
plt.xlabel('Policy Age')
plt.ylabel('Missingness')
plt.legend()

plt.show()
```

We see this behaviour in the majority cases for group filters. This suggests that the missingness is likely an issue to be resolved in the dataset rather than a "feature". With Length values being added as the policy persists.

Due to this, ML for data imputation can be seen as a reasonable solution to this issue.

```python
mask = (
    (df_plotting['y'] > 0) &
    np.isfinite(df_plotting['x']) &
    np.isfinite(df_plotting['y'])
)

x = df_plotting.loc[mask, 'x']
log_y = np.log(df_plotting.loc[mask, 'y'])

m, c = np.polyfit(x, log_y, 1)

a = np.exp(c)

x_smooth = np.linspace(df_plotting['x'].min(), df_plotting['x'].max(), 100)
y_pred = a * np.exp(m * x_smooth)


# 1. Log-Scale R² (R² of the linear fit on log_y)
log_y_pred = m * x + c
ss_res_log = np.sum((log_y - log_y_pred) ** 2)
ss_tot_log = np.sum((log_y - np.mean(log_y)) ** 2)
r2_log = 1 - (ss_res_log / ss_tot_log)

# 2. Original-Scale R² (R² of the exponential curve on the actual y values)
# We calculate predictions for ALL data points (including zeros) to be fair
y_pred_all = a * np.exp(m * df_plotting['x'])
ss_res_all = np.sum((df_plotting['y'] - y_pred_all) ** 2)
ss_tot_all = np.sum((df_plotting['y'] - df_plotting['y'].mean()) ** 2)
r2_original = 1 - (ss_res_all / ss_tot_all)

print(f"Log-Scale R²: {r2_log:.4f}")
print(f"Original-Scale R²: {r2_original:.4f}")
```

```python
# 1. Get predictions for both models
y_pred_exp = a * np.exp(m * df_plotting['x'])       # Exponential predictions
y_pred_nb = nb_model.predict(df_plotting)           # Neg Binomial predictions

# 2. Calculate and print metrics

print(f"Neg Binomial - MAE: {mean_absolute_error(df_plotting['y'], y_pred_nb):.2f} | RMSE: {np.sqrt(mean_squared_error(df_plotting['y'], y_pred_nb)):.2f}")
print(f"Exponential  - MAE: {mean_absolute_error(df_plotting['y'], y_pred_exp):.2f} | RMSE: {np.sqrt(mean_squared_error(df_plotting['y'], y_pred_exp)):.2f}")
```

```python
gbr = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),

    ("model", GradientBoostingRegressor(
        n_estimators=200,
        learning_rate=0.1,
        loss="absolute_error", # We want absolute error to avoid the impact the squared_error has on GBR
        max_depth=3,
        random_state=42
    ))
])

# parameter options for the tunning
param_distributions = {
    "model__n_estimators": [100, 200, 300],
    "model__learning_rate": [0.03, 0.05, 0.1],
    "model__max_depth": [2, 3, 4],
    "model__min_samples_leaf": [1, 3, 5]
}

gbr_search = RandomizedSearchCV(
    estimator=gbr,
    param_distributions=param_distributions,
    n_iter=20,
    scoring=scoring,
    refit= 'MAE',
    cv=cv,
    #groups=groups,
    n_jobs=N_jobs,
    random_state=42,
    verbose=2
)


# Fit on all known Length observations
gbr_search.fit(X, y,groups=groups)

print(gbr_search.best_params_)
print(-gbr_search.best_score_)

gbr_best = gbr_search.best_estimator_

results = gbr_search.cv_results_
best_idx = gbr_search.best_index_

# Validation scores
mae = -results["mean_test_MAE"][best_idx]
mae_std = results["std_test_MAE"][best_idx]

r2 = results["mean_test_R2"][best_idx]
r2_std = results["std_test_R2"][best_idx]

rmse = -results["mean_test_RMSE"][best_idx]
rmse_std = results["std_test_RMSE"][best_idx]

print(f"Mean MAE: {mae:.4f}")
print(f"MAE standard deviation: {mae_std:.4f}")

print(f"Mean R^2:  {r2:.4f}")
print(f"R^2 standard deviation: {r2_std:.4f}")

print(f"Mean RMSE: {rmse:.4f}")
print(f"RMSE standard deviation: {rmse_std:.4f}")

# Preserve the original Length column
Curated_Lengths["Length_GBR"] = Length_imputation["Length"]

Curated_Lengths.loc[missing_mask, "Length_GBR"] = gbr_best.predict(
    Length_imputation.loc[missing_mask, features]
)
```

These results suggest that the RFR is very effective for imputing the lengths data. While the GBR lags behind. This is due to GBR requiring tunning of parameters for effective usage.
